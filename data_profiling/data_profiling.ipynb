```python
import requests
import pandas as pd
pd.set_option('display.max.columns', None)
```

```python
def return_response(uri):
    r = requests.get(uri)
    return r.json()
```

```python
def return_df(uri):
    
    json_data = return_response(uri)
    
    overview = pd.json_normalize(json_data).drop("network.stations", axis = 1)
    
    inner_net_specific_overview = pd.json_normalize(json_data['network']['stations'])

    return pd.merge(overview, inner_net_specific_overview, how = 'cross')
```

```python
net_specific_uri_v1 = 'https://api.citybik.es/v2/networks/acces-velo-saguenay'
net_specific_uri_v2 = 'https://api.citybik.es/v2/networks/abu-dhabi-careem-bike'

net_general_uri = 'https://api.citybik.es/v2/networks/'
```

```python
net_specific_overview_1 = return_df(net_specific_uri_v1)
net_specific_overview_1[:2]
```

```python
net_specific_overview_2 = return_df(net_specific_uri_v2)
net_specific_overview_2[:2]
```

```python
json_data = return_response(net_general_uri)

df_all = pd.json_normalize(json_data['networks'])
df_all[:2]
```
